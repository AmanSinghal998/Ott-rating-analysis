```python
# OTT Rating Analysis - basic Python (no external libraries)
import csv

PLATFORMS = {"N": "Netflix", "P": "Prime Video", "H": "Hotstar",
             "M": "HBO / Max", "A": "Apple TV+", "S": "SonyLIV"}

# title | type | platform | year | genre | rating
raw = """Stranger Things|S|N|2016|Sci-Fi|8.7
Breaking Bad|S|N|2008|Crime|9.5
The Crown|S|N|2016|Drama|8.6
Squid Game|S|N|2021|Thriller|8.0
Money Heist|S|N|2017|Crime|8.2
Narcos|S|N|2015|Crime|8.8
Ozark|S|N|2017|Crime|8.4
Black Mirror|S|N|2011|Sci-Fi|8.7
The Irishman|M|N|2019|Crime|7.8
Roma|M|N|2018|Drama|7.7
Wednesday|S|N|2022|Fantasy|8.1
Dark|S|N|2017|Sci-Fi|8.7
Sacred Games|S|N|2018|Crime|8.5
Delhi Crime|S|N|2019|Crime|8.5
Bridgerton|S|N|2020|Romance|7.3
The Witcher|S|N|2019|Fantasy|8.0
Lupin|S|N|2021|Crime|7.5
Adolescence|S|N|2025|Drama|8.1
Kota Factory|S|N|2019|Drama|9.0
The Boys|S|P|2019|Action|8.7
Mirzapur|S|P|2018|Crime|8.4
The Family Man|S|P|2019|Action|8.7
Panchayat|S|P|2020|Comedy|8.9
Invincible|S|P|2021|Animation|8.7
Fleabag|S|P|2016|Comedy|8.7
Mrs. Maisel|S|P|2017|Comedy|8.7
Jack Ryan|S|P|2018|Action|8.0
Paatal Lok|S|P|2020|Crime|8.1
Made in Heaven|S|P|2019|Drama|7.9
The Mandalorian|S|H|2019|Sci-Fi|8.7
Loki|S|H|2021|Fantasy|8.2
WandaVision|S|H|2021|Fantasy|7.9
Special Ops|S|H|2020|Thriller|8.6
Andor|S|H|2022|Sci-Fi|8.4
Aarya|S|H|2020|Crime|7.6
Avengers: Endgame|M|H|2019|Action|8.4
Game of Thrones|S|M|2011|Fantasy|9.2
Chernobyl|S|M|2019|Drama|9.3
The Last of Us|S|M|2023|Drama|8.7
Succession|S|M|2018|Drama|8.8
House of the Dragon|S|M|2022|Fantasy|8.4
The Wire|S|M|2002|Crime|9.3
Ted Lasso|S|A|2020|Comedy|8.8
Severance|S|A|2022|Sci-Fi|8.7
CODA|M|A|2021|Drama|8.0
Slow Horses|S|A|2022|Thriller|8.1
Scam 1992|S|S|2020|Drama|9.2
Gullak|S|S|2019|Comedy|9.1
Rocket Boys|S|S|2022|Drama|8.9
Maharani|S|S|2021|Drama|7.8
Peaky Blinders|S|N|2013|Crime|8.7
The Queen’s Gambit|S|N|2020|Drama|8.6
Mindhunter|S|N|2017|Crime|8.6
Narcos: Mexico|S|N|2018|Crime|8.3
Cobra Kai|S|N|2018|Action|8.5
Heartstopper|S|N|2022|Romance|8.7
Sex Education|S|N|2019|Comedy|8.3
Arcane|S|N|2021|Animation|9.0
Beef|S|N|2023|Comedy|8.2
Jamtara|S|N|2020|Crime|7.6
Heeramandi|S|N|2024|Drama|6.9
Rana Naidu|S|N|2023|Crime|6.9
Reacher|S|P|2022|Action|8.1
Fallout|S|P|2024|Sci-Fi|8.3
Bandish Bandits|S|P|2020|Drama|8.1
Inside Edge|S|P|2017|Drama|7.8
The Wheel of Time|S|P|2021|Fantasy|7.1
Dahaad|S|P|2023|Crime|7.9
Farzi|S|P|2023|Thriller|8.4
Breathe|S|P|2018|Thriller|8.3
Good Omens|S|P|2019|Fantasy|8.0
Hunters|S|P|2020|Thriller|7.3
Jubilee|S|P|2023|Drama|8.2
Criminal Justice|S|H|2019|Crime|8.0
Hostages|S|H|2019|Thriller|7.8
Shogun|S|H|2024|Drama|8.6
The Bear|S|H|2022|Comedy|8.6
Ahsoka|S|H|2023|Sci-Fi|7.3
Moon Knight|S|H|2022|Action|7.3
Avatar|M|H|2009|Sci-Fi|7.9
Hawkeye|S|H|2021|Action|7.5
Euphoria|S|M|2019|Drama|8.3
Band of Brothers|S|M|2001|Drama|9.4
The Sopranos|S|M|1999|Crime|9.2
True Detective|S|M|2014|Crime|8.9
Westworld|S|M|2016|Sci-Fi|8.5
Barry|S|M|2018|Comedy|8.4
The White Lotus|S|M|2021|Comedy|7.9
Dune|M|M|2021|Sci-Fi|8.0
The Morning Show|S|A|2019|Drama|8.0
For All Mankind|S|A|2019|Sci-Fi|8.0
Foundation|S|A|2021|Sci-Fi|7.5
Shrinking|S|A|2023|Comedy|8.1
Silo|S|A|2023|Sci-Fi|8.2
Pachinko|S|A|2022|Drama|8.2
Killers of the Flower Moon|M|A|2023|Crime|7.6
Avrodh|S|S|2020|Thriller|8.5
Tanaav|S|S|2022|Thriller|8.4
Scam 2003|S|S|2023|Crime|8.3
Undekhi|S|S|2020|Crime|7.5"""

# 1. Load the data into a list of dictionaries
movies = []
for line in raw.strip().split("\n"):
    title, kind, code, year, genre, rating = line.split("|")
    movies.append({
        "title": title,
        "type": "Movie" if kind == "M" else "Show",
        "platform": PLATFORMS[code],
        "year": int(year),
        "genre": genre,
        "rating": float(rating),
    })

# 2. Basic statistics
total = len(movies)
average = sum(m["rating"] for m in movies) / total
print("Total titles  :", total)
print("Average rating:", round(average, 2))

# 3. Average rating by platform (dictionary + loop)
platform_ratings = {}
for m in movies:
    platform_ratings.setdefault(m["platform"], []).append(m["rating"])

print("\nAverage rating by platform")
platform_avg = {p: round(sum(r) / len(r), 2) for p, r in platform_ratings.items()}
for p, avg in sorted(platform_avg.items(), key=lambda x: x[1], reverse=True):
    print(f"{p:12} {avg:5}  {'#' * int((avg - 6) * 10)}")   # text bar chart

# 4. Titles per genre
genre_count = {}
for m in movies:
    genre_count[m["genre"]] = genre_count.get(m["genre"], 0) + 1

print("\nTitles by genre")
for g, c in sorted(genre_count.items(), key=lambda x: x[1], reverse=True):
    print(f"{g:12} {c}")

# 5. Top 5 rated titles
print("\nTop 5 rated titles")
top5 = sorted(movies, key=lambda m: m["rating"], reverse=True)[:5]
for i, m in enumerate(top5, 1):
    print(f"{i}. {m['title']} ({m['platform']}) - {m['rating']}")

# 6. Simple search (like the website's search box)
keyword = input("\nSearch title or genre: ").lower()
for m in movies:
    if keyword in m["title"].lower() or keyword in m["genre"].lower():
        print(m["title"], "|", m["platform"], "|", m["rating"])

# 7. Save the dataset as CSV using the built-in csv module
with open("ott_dataset.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=movies[0].keys())
    writer.writeheader()
    writer.writerows(movies)
print("\nSaved ott_dataset.csv")
```
